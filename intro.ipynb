```python
learning_rate = 0.1
x = torch.tensor(5.0, requires_grad=True)
for iterator in range(100):
    f = x ** 2 # forward pass
    f.backward() # backward pass
    with torch.no_grad(): # tells pytorch not to record this operation 
        x -= learning_rate * x.grad # gradient descent step

    x.grad.zero_() # reset the gradients for the new pass
```
