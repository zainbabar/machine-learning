## MLP for regression

```python
from sklearn.datasets import fetch_california_housing
from sklearn.metrics import root_mean_squared_error
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPRegressor
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
```

```python
housing = fetch_california_housing()
X_train, X_test, y_train, y_test = train_test_split(housing.data, housing.target, random_state=42)
```

create a neural net w/ 3 hidden layers w/ 50 neurons each. input & output layers adjust automatically based on our inputs and targets when training starts. 

model uses reLU activation function for the hidden layers and no activation function on the output layer (don't wanna squash prices)

early stopping stops the training if the model doesnt improve for 10 epochs straight (default # of epochs) tested on a subset of the training data

```python
mlp_reg = MLPRegressor(hidden_layer_sizes=[50,50,50], early_stopping=True, verbose=True, random_state=42)
```

need to standardize the input features before passing then into the nn, since gradient descent doesnt converge well when features have diff scales
- diff scales for the features results in diff scales for the weights 
- i.e. a feature from 100-200k vs a feature on 0-1. large features gradient is huge, cuz it affects the neurons output way more just cuz of its magnitude. 
- to optimize the gradients we need one learning rate that works for both
    - way too much time with a low learning rate, makes very slow progress
    - high learning rate, messes up the gradient update of the smaller feature, can oscillate and overshoot and diverge

standardizing puts all features on similar scales, so gradients are comparable, and a single learning rate works well for all

create a pipeline to scale the data and pass through into the model

```python
pipeline = make_pipeline(StandardScaler(), mlp_reg)
```

```python
pipeline.fit(X_train, y_train)
```

```python
mlp_reg.best_validation_score_ # note that score ended up going down over the last few epocs
```

```python
y_pred = pipeline.predict(X_test) # uses the best model not the last model
```

```python
rmse = root_mean_squared_error(y_test, y_pred)
rmse
```

red line is true values, dots are predictions

```python
import matplotlib.pyplot as plt
plt.scatter(y_test, y_pred, alpha=0.1)
plt.axline((0,0), slope=1, color='red')
plt.show()
```
